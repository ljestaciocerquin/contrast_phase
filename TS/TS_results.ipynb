```python
def show_slices_per_class(images, center=50, width=300):

    def window_ct(img, center, width):
        low = center - width/2
        high = center + width/2
        return img.clip(low, high)


    class_ranges = {
        "Arterial": range(0, 3),
        "Non-contrast": range(3, 6),
        "Portal": range(6, 9)
    }

    for cls_name, indices in class_ranges.items():

        if cls_name == 'Portal':
            z = 50
        else:
            z = int(np.mean([images[i].shape[3] for i in indices]))- 30

        plt.figure(figsize=(15,5))
        for j, i in enumerate(indices):
            img = images[i][0, :, :, z]
            plt.subplot(1, 3, j+1)
            plt.imshow(np.rot90(window_ct(img, center, width),k=1), cmap="gray")
            plt.title(f"{cls_name} image {i}, slice {z}")
            plt.axis("off")
        plt.show()

show_slices_per_class(misclass_images)
```

```python
def estimate_time_interval(paired_data):

    pairs = []

    # Per exam
    for (patient_id, exam_date), group in paired_data.groupby(
        ["SubjectKeyRadiology", "ExamDate"]
    ):

        group = group.sort_values("AcquisitionTime_sec")

        arterial = group[group["contrast"] == "Arterial"]
        portal = group[group["contrast"] == "Portal"]

        if arterial.empty or portal.empty:
            continue

        # For each arterial, pair with ALL future portal variants

        for _, a_row in arterial.iterrows():

            a_time = a_row["AcquisitionTime_sec"]
            a_pi_time = a_row["pi_time"]
            future_portals = portal[portal["AcquisitionTime_sec"] > a_time]

            if future_portals.empty:
                continue

            a_phase = a_row["phase_timing"]

            # Preserve multiple kernel reconstructions

            for _, p_row in future_portals.iterrows():

                p_time = p_row["AcquisitionTime_sec"]
                p_pi_time = p_row["pi_time"]
                interval = p_time - a_time
                TS_interval = p_pi_time - a_pi_time

                if interval <= 0 or TS_interval <= 0:
                    continue

                p_phase = p_row["phase_timing"]
                pairs.append({

                    "SubjectKeyRadiology": patient_id,
                    "ExamDate": exam_date,
        
                    "arterial_timing": a_phase,
                    "portal_timing": p_phase,

                    # DICOM acquisition time interval
                    "time_interval": interval,

                    # TS predicted time interval
                    "TS_time_interval": TS_interval

                })

    return pd.DataFrame(pairs)
```

```python
time_interval_results =  estimate_time_interval(contrast_data)
time_interval_results
```

```python
# time_interval_results.time_interval.info()

from sklearn.metrics import root_mean_squared_error, mean_absolute_error

print("MAE:", mean_absolute_error(time_interval_results.time_interval,time_interval_results.TS_time_interval))
print("RMSE:", root_mean_squared_error(time_interval_results.time_interval,time_interval_results.TS_time_interval))
```
