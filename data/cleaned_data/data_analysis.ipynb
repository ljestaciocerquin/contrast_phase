```python
import matplotlib.pyplot as plt
import seaborn as sns

def data_stats(data):
    print('Number of patients:', data['SubjectKeyRadiology'].nunique(), 
      '\nNumber of files:', data['Original File'].nunique(), 
      '\nNumber of patients with both arterial and portal:', data[data.contrast.isin(['Arterial', 'Portal'])]
                                                                                .groupby('SubjectKeyRadiology')['contrast']
                                                                                .nunique()
                                                                                .eq(2)
                                                                                .sum()
                                                                                )
    

def time_distribution(data):
    # Remove missing values if any
    time_data = data["time_interval"].dropna()

    fig, axes = plt.subplots(1, 2, figsize=(12, 4))

    # KDE plot
    sns.kdeplot(time_data, fill=True, ax=axes[0])
    axes[0].set_title("KDE of Time Interval")
    axes[0].set_xlabel("Time Interval")

    # Box plot
    sns.boxplot(x=time_data, ax=axes[1])
    axes[1].set_title("Box Plot of Time Interval")
    axes[1].set_xlabel("Time Interval")

    plt.tight_layout()
    plt.show()
    
```

```python
data_stats(data)
```

```python
time_data = time_interval_data['time_interval']
plt.figure(figsize=(8, 4))
sns.histplot(time_data, kde=True, bins=80, color="indianred")
plt.title("Distribution of Time Interval")
plt.xlabel("Time Interval")
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(time_data, bins=80, kde=True, color="indianred")
plt.xlim(0, time_data.quantile(0.99))  # show first 99% of values
plt.title("Time Interval Distribution (without extreme outliers)")
plt.xlabel("Time Interval")
plt.show()
```

```python
# Boxplot
plt.figure(figsize=(8, 5))

sns.boxplot(x=time_data)

plt.title("Boxplot of Time Interval", fontsize=16)
plt.xlabel("Time Interval")

plt.tight_layout()
plt.show()
```

```python
print(time_data.describe())

# IQR-based outlier counts
Q1 = time_data.quantile(0.25)
Q3 = time_data.quantile(0.75)
IQR = Q3 - Q1

outliers = time_data[
    (time_data < Q1 - 1.5 * IQR) |
    (time_data > Q3 + 1.5 * IQR)
]

print(f"Number of outliers: {len(outliers)}")
```
