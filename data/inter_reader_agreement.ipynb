```python
giovanni_batch3 = giovanni_batch3.rename(columns={'no_lesions': 'is_lesionfree'})
christina_batch3 = christina_batch3.rename(columns={'no_lesions': 'is_lesionfree'})
# kalina_batch3 = kalina_batch3.rename(columns={'no_lesions': 'is_lesionfree'})


def annot_mapping(data):
    contrast_mapping = {1.0: "Arterial", 2.0: "Portal", 3.0: "Non-contrast"}                       
    time_mapping = {1.0: "Too Early", 2.0: "Just Right", 3.0: "Too Late", 0.0: 0.0}                 
    liver_mapping = {1.0: "Yes", 2.0: "Partially"}                                                  
    lesion_mapping = {True: "Yes", False: "No"}

    data['contrast'] = data['contrast'].map(contrast_mapping)
    data['phase_timing'] = data['phase_timing'].map(time_mapping)
    data['is_liver_imaged'] = data['is_liver_imaged'].map(liver_mapping)
    data.loc[data['file'].notna(), 'is_lesionfree'] = data.loc[data['file'].notna(), 'is_lesionfree'].map(lesion_mapping).fillna("No")

    return data
```

```python
def contrast_timing(dataset):
    dataset["contrast_timing"] = dataset["contrast"]
    mask = dataset["contrast"] != "Non-contrast"
    dataset.loc[mask, "contrast_timing"] = (dataset.loc[mask, "contrast"] + " " + dataset.loc[mask, "phase_timing"].astype(str))
    dataset.loc[mask, "contrast_timing"] = dataset.loc[mask, "contrast_timing"].str.strip()
    dataset.loc[dataset["contrast"] == "Non-contrast", "contrast_timing"] = "Non-contrast"   

    return dataset
```

```python
giovanni_batch3 = contrast_timing(giovanni_batch3)
christina_batch3 = contrast_timing(christina_batch3)
batch3 = contrast_timing(batch3)
```

```python
import seaborn as sns
from sklearn.metrics import cohen_kappa_score
import matplotlib.pyplot as plt


def eval_IAA(data1, data2, key="file", label="contrast",
             suffixes=("_kalina", "_giovanni")):

    merged = pd.merge(data1, data2, on=key, suffixes=suffixes)
    agreement = (
        merged[f"{label}{suffixes[0]}"] ==
        merged[f"{label}{suffixes[1]}"]
    ).mean()

    return agreement,len(merged)

def eval_cohen(data1, data2, key="file", label="contrast",
               suffixes=("_kalina", "_giovanni")):

    merged = pd.merge(data1, data2, on=key, suffixes=suffixes)

    print("Number of overlapping samples:", len(merged))

    col1 = f"{label}{suffixes[0]}"
    col2 = f"{label}{suffixes[1]}"

    kappa = cohen_kappa_score(merged[col1], merged[col2])

    print("Cohen's Kappa:", round(kappa,3))

    crosstab = pd.crosstab(
        merged[col1],
        merged[col2],
        normalize="all"
    )

    print(f"\nCrosstab of {label} annotations:")
    display(crosstab)

    return crosstab
```

```python
IAA_kg, n = eval_IAA(batch3, giovanni_batch3, label = "contrast", suffixes = ("_kalina", "_giovanni"))
print(f"Agreement: {IAA_kg:.2%} ({n} files)")
eval_cohen(batch3, giovanni_batch3, label = "contrast", suffixes = ("_kalina", "_giovanni"))


IAA_kg, n = eval_IAA(batch3, giovanni_batch3, label = "contrast_timing", suffixes = ("_kalina", "_giovanni"))
print(f"Agreement: {IAA_kg:.2%} ({n} files)")
cross_tab_kg = eval_cohen(batch3, giovanni_batch3, label = "contrast_timing", suffixes = ("_kalina", "_giovanni"))

sns.heatmap(
    cross_tab_kg,
    annot=True,
    fmt=".2f",
    cmap="Blues",
    square=True,
    cbar_kws={"label": "Proportion of all samples"}

    )

plt.xlabel("A1")
plt.ylabel("A3")
plt.title(f"Normalized crosstab of phase timing")
plt.tight_layout()
plt.show()
```
