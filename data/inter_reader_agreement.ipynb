```python
import pandas as pd
```

```python
data = pd.read_csv("/projects/net_contrast_classification/contrast_phase/data/cleaned_data/cleaned_data.csv")
batch3 = data[data["Destination Folder"].str.contains("DICOM-batch3")].reset_index(drop=True)
batch3
```

```python
giovanni_batch3 = pd.read_csv("/projects/net_contrast_classification/contrast_phase/data/giovanni_data/NET_annotations.csv")
christina_batch3 = pd.read_csv("/projects/net_contrast_classification/contrast_phase/data/christina_data/NET_annotations.csv")

print(len(giovanni_batch3), len(christina_batch3))
```

```python
print("Overlap between Giovanni and Christina:",len(set(giovanni_batch3["file"]).intersection(christina_batch3["file"])), "samples")
print("Overlap between Giovanni and Kalina:",len(set(giovanni_batch3["file"]).intersection(batch3["file"])), "samples")
print("Overlap between Christina and Kalina:",len(set(batch3["file"]).intersection(christina_batch3["file"])), "samples")

```

```python
giovanni_batch3 = giovanni_batch3.rename(columns={'no_lesions': 'is_lesionfree'})
christina_batch3 = christina_batch3.rename(columns={'no_lesions': 'is_lesionfree'})

def annot_mapping(data):
    contrast_mapping = {1.0: "Arterial", 2.0: "Portal", 3.0: "Non-contrast"}                       
    time_mapping = {1.0: "Too Early", 2.0: "Just Right", 3.0: "Too Late", 0.0: 0.0}                 
    liver_mapping = {1.0: "Yes", 2.0: "Partially"}                                                  
    lesion_mapping = {True: "Yes", False: "No"}

    data['contrast'] = data['contrast'].map(contrast_mapping)
    data['phase_timing'] = data['phase_timing'].map(time_mapping)
    data['is_liver_imaged'] = data['is_liver_imaged'].map(liver_mapping)
    data.loc[data['file'].notna(), 'is_lesionfree'] = data.loc[data['file'].notna(), 'is_lesionfree'].map(lesion_mapping).fillna("No")

    return data
```

```python
display(giovanni_batch3.contrast.value_counts(), giovanni_batch3.contrast.isna().sum(),
        giovanni_batch3.phase_timing.value_counts(), giovanni_batch3.phase_timing.isna().sum(),
        giovanni_batch3.is_liver_imaged.value_counts(), giovanni_batch3.is_liver_imaged.isna().sum(),
        giovanni_batch3.is_lesionfree.value_counts(), giovanni_batch3.is_lesionfree.isna().sum())
```

```python
giovanni_batch3 = giovanni_batch3.rename(columns={'no_lesions': 'is_lesionfree'})
christina_batch3 = christina_batch3.rename(columns={'no_lesions': 'is_lesionfree'})

giovanni_batch3 = annot_mapping(giovanni_batch3)
christina_batch3 = annot_mapping(christina_batch3)
```

```python
print(
    "Duplicate files in Giovanni:",
    giovanni_batch3["file"].duplicated().sum()
)

print(
    "Duplicate files in Christina:",
    christina_batch3["file"].duplicated().sum()
)

print(
    "Duplicate files in Kalina:",
    batch3["file"].duplicated().sum()
)
```

```python
g_dups = giovanni_batch3[
    giovanni_batch3["file"].duplicated(keep=False)
].sort_values("file")

c_dups = christina_batch3[
    christina_batch3["file"].duplicated(keep=False)
].sort_values("file")

k_dups = batch3[
    batch3["file"].duplicated(keep=False)
].sort_values("file")


display(g_dups)
display(c_dups)
display(k_dups)
```

```python
giovanni_batch3 = giovanni_batch3.drop_duplicates(subset="file", keep="first")
christina_batch3 = christina_batch3.drop_duplicates(subset="file", keep="first")
```

```python
overlap_gc = set(giovanni_batch3["file"]).intersection(christina_batch3["file"])

gc = (
    giovanni_batch3[["file", "contrast"]]
    .merge(
        christina_batch3[["file", "contrast"]],
        on="file",
        suffixes=("_giovanni", "_christina")
    )
)

print(len(overlap_gc))
print("Number of overlapping samples:", len(gc))
```

```python
from sklearn.metrics import cohen_kappa_score

kappa = cohen_kappa_score(
    gc["contrast_giovanni"],
    gc["contrast_christina"]
)

print(f"Giovanni-Christina Cohen's Kappa: {kappa:.3f}")
pd.crosstab(
    gc["contrast_giovanni"],
    gc["contrast_christina"],
    normalize="all"
)
```

```python
gk = (
    batch3[["file", "contrast"]]
    .merge(
        giovanni_batch3[["file", "contrast"]],
        on="file",
        suffixes=("_kalina", "_giovanni", )
    )
)

kappa_gk = cohen_kappa_score(
    gk["contrast_giovanni"],
    gk["contrast_kalina"]
)

print(f"Giovanni-Kalina Cohen's Kappa: {kappa_gk:.3f}")
pd.crosstab(
    gk["contrast_giovanni"],
    gk["contrast_kalina"],
    normalize="all"
)
```

```python
ck = (
    batch3[["file", "contrast"]]
    .merge(
        christina_batch3[["file", "contrast"]],
        on="file",
        suffixes=("_kalina", "_christina")
    )
)

kappa_ck = cohen_kappa_score(
    ck["contrast_christina"],
    ck["contrast_kalina"]
)

print(f"Kalina-Christina Cohen's Kappa: {kappa_ck:.3f}")
pd.crosstab(
    ck["contrast_christina"],
    ck["contrast_kalina"],
    normalize="all"
)
```

```python

ck_mismatches = ck[ck["contrast_christina"] != ck["contrast_kalina"]]
gk_mismatches = gk[gk["contrast_giovanni"] != gk["contrast_kalina"]]

print(ck_mismatches["contrast_kalina"].value_counts())
print(gk_mismatches["contrast_kalina"].value_counts())
```
