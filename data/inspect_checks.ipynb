```python
import pandas as pd
```

```python
zeros_checked = pd.read_csv("/projects/net_contrast_classification/contrast_phase/data/kalina_checks/checked/kalina_check_zeros_checked.csv", delimiter = ";")
data = pd.read_csv("/projects/net_contrast_classification/contrast_phase/data/cleaned_data/cleaned_data.csv")

data.contrast.value_counts()
```

```python
zeros_checked = zeros_checked.drop('Unnamed: 6', axis = 1)
zeros_checked.info()
```

```python
print(zeros_checked.contrast.value_counts())
zeros_checked = zeros_checked[~zeros_checked.contrast.isin(['delayed', "0"])]
zeros_checked['contrast'] = zeros_checked['contrast'].apply(lambda x: "Non-contrast" if x == "no contrast" else x)
print(zeros_checked.contrast.value_counts())
```

```python
print(zeros_checked.phase_timing.value_counts())
zeros_checked = zeros_checked[zeros_checked.phase_timing != "0.0"]
zeros_checked['phase_timing'] = zeros_checked['phase_timing'].apply(lambda x: "Too Early" if x == "too early" else x)
zeros_checked['phase_timing'] = zeros_checked['phase_timing'].apply(lambda x: "Too Late" if x == "too late" else x)

print(zeros_checked.phase_timing.value_counts())
```

```python
print(zeros_checked.is_liver_imaged.value_counts())
zeros_checked['is_liver_imaged'] = zeros_checked['is_liver_imaged'].apply(lambda x: "Yes" if x == "yes" else x)

zeros_checked = zeros_checked[zeros_checked['is_liver_imaged']!= "exclude - coronal"]
print(zeros_checked.is_liver_imaged.value_counts())

```

```python
zeros_checked.to_csv("/projects/net_contrast_classification/contrast_phase/data/kalina_checks/checked/zeros_filtered.csv")
```

```python
cols_to_update = [c for c in zeros_checked.columns if c != "MatchKey"]
cols_to_update
```

```python
data = data.set_index("MatchKey")
zeros_checked = zeros_checked.set_index("MatchKey")

# Update matching rows and columns
data.update(zeros_checked[cols_to_update])

# Restore MatchKey as a column
data = data.reset_index()
data.contrast.value_counts()
```

```python
data = data[data.contrast != "0"]
data = data[data.phase_timing != "0.0"]

data.contrast.value_counts()
data.phase_timing.value_counts()
data.is_liver_imaged.value_counts()

data = data.reset_index(drop=True)
data.info()
```

```python
data = data[~data.is_liver_imaged.isna()].reset_index(drop=True)
```
